### Lists in Python: The Fundamentals
A Python list is a built-in, ordered, mutable, and dynamic collection that can store heterogeneous (mixed) data types.
### List Comprehensions
List comprehensions provide a syntactic shorthand to create new lists from existing iterables. They are typically faster than standard for loops because they are optimized at the C-level in CPython.
**Syntax:** [expression for item in iterable if condition]
#### when to use list 
 **Ideal Scenarios:** 
 1.Sequential Data Processing: When the sequence and order of arrival matter.  
 2.Frequent Access by Index: Accessing an element by its index (list[i]) is an $O(1)$ constant time operation.  
 3.Dynamic Stacks: Because append() and pop() operate on the end of the list in $O(1)$ time, a list works beautifully as a Last-In, First-Out (LIFO) stack


## Task 1: Student Grade Manager 


```python
students_name = ["Fatima","Ayesha","Muqadas","Hina"]
students_grade = [90, 85, 95, 80]


def calculate_average_grade(grades):
    addition  = sum(grades)
    average = addition / len(grades)
    return average

def add_student(name, grade):
    students_name.append(name)
    students_grade.append(grade)

def remove_student(name):
    if name in students_name:
        index = students_name.index(name)
        students_name.pop(index)
        students_grade.pop(index)
    else:
        print(f"Student with name {name} not found")
def update_grade(index,new_grade):
    students_grade[index]=new_grade

update_grade(2, 70)
print(students_grade)
# find highest and lowest grade
highest_grade =0
lowest_grade =0
def find_highest_grade():
    highest_grade =0
    highest_students_name = ""
    for x in students_grade:
        if x>highest_grade:
            highest_grade = x
            index = students_grade.index(x)
            highest_students_name= students_name[index]
    print(f"{highest_students_name} has highest scroce grade {highest_grade}")

print(find_highest_grade())

def find_lowest_grade():
      lowest_grade = students_grade[0]
      lower_grade_student = ""
      for x in students_grade:
        if x<lowest_grade:
            lowest_grade = x
            index = students_grade.index(x)
            lower_grade_student= students_name[index]
      print(f"{lower_grade_student} has lower scroce grade {lowest_grade}")


print(find_lowest_grade())

def filter_students_above_avg():
    avg = calculate_average_grade(students_grade)
    above_avg_students = [students_name[i] for i in range(len(students_grade)) if students_grade[i] > avg]
    return above_avg_students
def filter_students_below_avg():
    avg = calculate_average_grade(students_grade)
    below_avg_students = [students_name[i]for i in range(len(students_grade)) if students_grade[i] < avg]
    return below_avg_students

# Sort students by grade (descending), display top 3 performers 
new_list = sorted(zip(students_name, students_grade), key=lambda x: x[1], reverse=True)
top_3_performers = new_list[:3]
print("Top 3 performers:")
for name, grade in top_3_performers:
    print(f"{name}: {grade}")
```

## Task 2: Shopping Cart System 


```python

items = ["Milk", "Bread", "Eggs", "Rice", "Sugar", "Tea"]
prices = [2.5, 1.5, 3.0, 20.0, 5.0, 4.0]
quantities = [2, 3, 1, 1, 2, 1]

def add_item(item, price,quantity):
    items.append(item)
    prices.append(price)
    quantities.append(quantity)
    return items,prices,quantities
def remove_item(item, price,quantity):
    items.remove(item)
    prices.remove(price)
    quantities.remove(quantity)
    return items,prices,quantities
def update_quantity(index,quantity):
    quantities[index]= quantity
    return quantities

def total():
    calculate = 0
    for x in range(len(prices)):
        multiply= prices[x]* quantities[x]
        calculate+= multiply

    return calculate
print("Add items", add_item("Oil",8.0,8))
print("Remove item", remove_item("Eggs",3.0,1))
print("Add items", add_item("Biscuit",8.0,8))
print("update quantity", update_quantity(4,7))
total_price = total()
print("Total Price", total_price)
if(total_price>100):
 discount =   total_price*0.10;
 discounted_price = total_price-discount
 print("discounted price",discounted_price)
else:
    "not discount"
print(items[-3:])
```
