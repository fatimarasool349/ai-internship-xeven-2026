## Supervised, unsupervised, and reinforcement learning with examples   
**Supervised Learning:**  
Supervised Learning is learning from labeled data. A label is the correct answer already attached to each example.
**Examples:**
Email Spam Detection  
**Un-Supervised Learning:**  
Unsupervised Learning learns from unlabeled data.No correct answers are given.The model must discover patterns itself.
**Examples:**  
Anomaly Detection:  Finding unusual data.   
**Reinforcement:**  
An agent learns by taking actions in an environment and receiving rewards or penalties. It improves over time by figuring out which actions lead to the best outcomes.  
**Example:**  
Teaching a Dog

**Semi-Supervised:**    
Uses both labeled and unlabeled data. Only small portion of data is labeled.  
**example:**  
A photo app has labels for only some images (e.g., "cat" or "dog") and uses the unlabeled images to improve its predictions.  
### Types of machine Learning (claud)
it also give two more type of machine learning Self-Learning and Transfer Learning  
**Self-Supervised:**  
the model learns by filling in blanks in its own data. No human labels needed; the data teaches itself.  
**Example:** Predicting the next word in a sentence.   
**Transfer Learning:**    
take a model already trained on one big task, and reuse it for a different, smaller task instead of starting from zero.   
**Example:**
Google's image model was trained on millions of everyday photos (cats, cars, trees). A hospital then takes that same model and shows it 500 X-ray images of lungs. It quickly learns to detect pneumonia — because it already understood shapes, edges, and patterns from the photos. Without transfer learning, the hospital would need hundreds of thousands of labeled X-rays.
