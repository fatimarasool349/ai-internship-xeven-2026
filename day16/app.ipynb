## What is LangChain?
LangChain is a Python framework that helps developers build applications using Large Language Models (LLMs) like GPT.
## Core Components of LangChain
1. **Models (LLMs):** A model is simply the AI. like chatgpt, llama
2. **Prompts (Templates):**  Instead of writing prompts every time, create a template.
3. **Chains (Workflow):** A chain connects multiple steps together.  
     Why is it called a Chain?  
     Because each step passes its output to the next step.   
     **Example**:  user Qurstion->Prompt-> LLM=> Paser-> output    
4. **Memory:** Nomally gpt forget you information without memory
## LCEL (LangChain Expression Language)
LCEL is a simple way to connect components using the | operator.
- Runnable Interface: Everything in LangChain is a Runnable. That means every component understands .invoke()
- Document Loaders: first exact text from document using PyPDFLoader() then .load() to load exacted text
## When to Use LangChain?

Use LangChain when you have more complex workflows, such as:

1. Building a chatbot with conversation history (memory).
2. Reading and querying PDFs or websites.
3. Connecting multiple LLM steps into a workflow (chains).
4. Combining prompts, models, document retrieval, and output parsing.
5. Building AI agents that can use tools (search, databases, calculators, etc.).

**processing, memory, or agents.** It provides higher-level building blocks so you can focus on the application logic instead of wiring everything together manually.

```python
from langchain_groq import ChatGroq
from dotenv import load_dotenv
import os
load_dotenv()
```

```python
model = ChatGroq(
    model="llama-3.3-70b-versatile",
    temperature=0,
    api_key=os.getenv("OPENAI_API_KEY"),


)
```

```python
response = model.invoke("What is AI?")
print(response.content)
```

```python
from langchain_core.prompts import PromptTemplate
prompt = PromptTemplate.from_template(
    "Explain {topic} and give {number} example"
)
```

```python
from langchain_core.output_parsers import StrOutputParser
chain = prompt | model | StrOutputParser()
result = chain.invoke(
    {
        "topic": "langchain", "number": 2
    }
)
print(result)
```

```python
from pathlib import Path

from langchain_community.document_loaders import (
    TextLoader,
    PyPDFLoader,
)

def load_document(file_path):
    extension = Path(file_path).suffix.lower()

    if extension == ".txt":
        loader = TextLoader(file_path)

    elif extension == ".pdf":
        loader = PyPDFLoader(file_path)

    else:
        raise ValueError(f"Unsupported file type: {extension}")

    return loader.load()

print(load_document("Topics.pdf"))
```
