**and:** it return true only if both operand are true.  
**or:** it return true if atleast one value is true.

## Comparison (Relational) operators:  
**==:** check both value equal or not **Use case:** Password checking  
**>=:** check first value is greater or equal. **Use Case:** conditional rendering  
**<=:** check first value is less or equal. **Use Cases:** conditional rendering  
**!=:** check both value not equal.  

### Arithmetic operators: +, -, *, /, // (floor division), % (modulus), ** (exponentiation) 

**+:** Adds two numbers together (e.g., 5 + 3 = 8).  
**-(Subtraction):** Subtracts the second number from the first (e.g., 8 - 3 = 5).  
*** (Multiplication):** Multiplies two numbers (e.g., 4 * 5 = 20).  
**/ (Division):** Divides one number by another and always returns a floating-point result (e.g., 10 / 4 = 2.5).  
**// (Floor Division):** Divides two numbers and returns the quotient rounded down to the nearest whole number (e.g., 10 // 4 = 2).  
**% (Modulus):** Returns the remainder after division (e.g., 10 % 4 = 2).  
****(Exponentiation):** Raises a number to the power of another number (e.g., 2 ** 3 = 8).  

### Operator precedence: understanding order of operations in Python  
Python's Priority List (Highest to Lowest)
1. () (Parentheses)
2.  (Exponentiation)
3. *, /, //, % (Multiplication/Divisions)
4. +, - (Addition/Subtraction)
5. <, >, <=, >=, ==, != (Comparisons)
6. not, and, or (Logical operators)

## Ask Gemini: 'What is operator precedence in Python and why does it matter?' 
In Python, when you write a single line of code that asks the computer to do multiple things—like add, multiply, and compare values—Python doesn't just read it left-to-right like a book. It looks at the symbols (operators), checks its internal rulebook for which symbol has the highest priority, and executes that one first.  
**Why It Matters (The Technical Risk)**  
If you don't understand precedence, your code will run without throwing an error, but it will calculate the completely wrong answer. These are called logic bugs, and they are the hardest to find because Python thinks it did exactly what you asked.

## Ask Claude: 'Best practices for type conversion and error handling in Python' 
1. Always convert input() immediately
2. Wrap risky conversions in try/except
3. Validate before converting when possible
4. Use or for safe defaults
5. Never swallow errors silently
