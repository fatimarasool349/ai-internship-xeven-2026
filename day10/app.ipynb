## Task 1: Student Information System 

```python
nested_dictionary = {101:
    {
    "name": "Fatima",
    "age": 22,
    "grades": {
        "cs": 98,
        "math": 76,
        "phy": 65, 
    }

},
    212:{
    "name":"Ayesha",
    "age": 23,
    "grades":{
        "cs": 87,
        "math":69,
        "phy":86
    }
}
}


    
    
     
    
```

```python
def add_student( student_id, name,age,cs,math,phy):
    nested_dictionary[student_id]={
            "name": name,
            "age": age,
            "grade":{
                "cs": cs,
                "math": math,
                "phy": phy
            }
        }
    
```

```python
student_id= int(input("Enter student ID:"))
student_name = input("Enter Student Name:")
student_age = int(input("Enter Student Age:"))
cs_grade = int(input("Enter cs grade:"))
math_grade = int(input("Enter math  grade:"))
phy_grade = int(input("Enter phy grade"))
    
add_student(student_id,student_name,student_age,cs_grade,math_grade,phy_grade)
print(nested_dictionary)
```

```python
def del_student(student_id):
    nested_dictionary.pop(student_id)
remove_student = int(input("Enter student id which you want to delete"))
del_student(remove_student)
print(nested_dictionary)
```

```python
def update_student_grade(student_id, cs_grade,math_grade,phy_grade):
    nested_dictionary[student_id]["grades"] = {
        "cs" : cs_grade,
        "math": math_grade,
        "phy": phy_grade
    }
student_id = int(input("Enter Student ID whose grade you want to update"))
cs_grade = int(input("Enter new grade"))
math_grade =  int(input("Enter new grade"))
phy_grade = int(input("Enter new grade"))

update_student_grade(student_id,cs_grade,math_grade,phy_grade)
print(nested_dictionary)
```

```python
def avg_of_student_grade(nested_dictionary):
    for student_id in nested_dictionary:
        grades=nested_dictionary[student_id]["grades"]
        avg = sum(grades.values()) / len(grades)
        print(avg)
avg_of_student_grade(nested_dictionary)

        
```

```python
def top_student(nested_dictionary):
    max =0
    for student_id in nested_dictionary:
        grades=nested_dictionary[student_id]["grades"]
        avg = sum(grades.values()) / len(grades)
        if(avg >max):
            max = avg
            higher_student_name = nested_dictionary[student_id]["name"]
    return higher_student_name    

top_student(nested_dictionary)
    
            
        
        
```

```python
# save and load stiudent in json

import json
with open("Student.json", "w") as file:
    json.dump(nested_dictionary,file, indent=4)
```

```python
# calculate GPA

def grade_point(mark):
     if mark >= 85:
        return 4.0
     elif mark >= 80:
        return 3.7
     elif mark >= 75:
        return 3.3
     elif mark >= 70:
        return 3.0
     elif mark >= 65:
        return 2.5
     elif mark >= 60:
        return 2.0
     elif mark >= 50:
        return 1.0
     else:
        return 0.0
    
for student_id in nested_dictionary:
    total =0
    for mark in nested_dictionary[student_id]["grades"].values():
        total += grade_point(mark)
    gpa = total/ len(nested_dictionary[student_id]["grades"])
    print(f"student name is  {nested_dictionary[student_id]['name'] } and gpa is {gpa}")
   
           
```

```python
products = {101:{
    "name": "Biryani Masala",
    "price": 200,
    "category": "Spices",
    "quantity": 2,
    
},
            
}
```

```python
def add_product(product_id,name,price,category,quantity):
    products[product_id]={
        "name": name,
        "price": price,
        "category": category,
        "quantity": quantity
        
    }
product_id= int(input("Enter produce ID:"))
product_name = input("Enter Product Name:")
product_price = input("Enter product price:")
product_category =(input("Enter category:"))
product_quantity = int(input("Enter product quantity:"))

add_product(product_id,product_name,int(product_price),product_category,product_quantity)

    
```

### Task 3: Configuration Manager 

```python
import json
with open("config.json", "r") as file:
    data = json.load(file)
    print(data) 
```

```python
def validate_config(data):
    required_keys = ["database", "api_keys", "features"]

    print("\nChecking configuration...")

    for key in required_keys:
        if key in data:
            print(f" {key} found")
        else:
            print(f" {key} is missing")
validate_config(data)
```

```python
def update_config(config):

    config["features"]["mode"] = "Light mode"
    config["features"]["notifications"] = False

    print("\nConfiguration Updated")
    
update_config(data)
```

```python
def save_config(config):

    with open("config.json", "w") as file:
        json.dump(config, file, indent=4)

    print("Configuration Saved Successfully")
    
save_config(data)
```
