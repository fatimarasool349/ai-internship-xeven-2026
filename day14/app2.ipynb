```python
import json
from collections import Counter
contacts = {}
```

```python
def add_contact():
    try:
        contact_id = int(input("Enter Contact ID: "))

        if contact_id in contacts:
            print("Contact ID already exists.")
            return

        name = input("Enter Name: ")
        phone = input("Enter Phone: ")
        email = input("Enter Email: ")

        tags = set(input("Enter Tags (comma separated): ").split(","))

        notes = input("Enter Notes (comma separated): ").split(",")

        contacts[contact_id] = {
            "name": name,
            "phone": phone,
            "email": email,
            "tags": tags,
            "notes": notes
        }

        print("Contact Added Successfully!")

    except ValueError:
        print("Invalid ID")
```

```python
def display_contacts():

    if not contacts:
        print("No Contacts Found")
        return

    for cid, detail in contacts.items():

        print("\n--------------------------")
        print("ID:", cid)
        print("Name:", detail["name"])
        print("Phone:", detail["phone"])
        print("Email:", detail["email"])
        print("Tags:", ", ".join(detail["tags"])) # .join use to merage string
        print("Notes:", ", ".join(detail["notes"]))
```

```python

def search_contacts():

    print("\n1. Search by Name")
    print("2. Search by Tag")
    print("3. Partial Name Search")

    choice = input("Choice: ")

    if choice == "1":

        name = input("Enter Name: ").lower()

        result = {
            cid: detail
            for cid, detail in contacts.items()
            if detail["name"].lower() == name
        }

    elif choice == "2":

        tag = input("Enter Tag: ").lower()

        result = {
            cid: detail
            for cid, detail in contacts.items()
            if tag in {t.lower() for t in detail["tags"]}
        }

    elif choice == "3":

        partial = input("Enter Partial Name: ").lower()

        result = {
            cid: detail
            for cid, detail in contacts.items()
            if partial in detail["name"].lower()
        }

    else:
        print("Invalid Choice")
        return

    if result:

        for cid, detail in result.items():

            print("\nID:", cid)
            print(detail)

    else:
        print("No Contact Found")
```

```python
def update_contact():

    cid = int(input("Enter Contact ID: "))

    if cid not in contacts:
        print("Contact Not Found")
        return

    print("Leave Blank to Keep Previous Value")

    name = input("Name: ")
    phone = input("Phone: ")
    email = input("Email: ")

    if name:
        contacts[cid]["name"] = name

    if phone:
        contacts[cid]["phone"] = phone

    if email:
        contacts[cid]["email"] = email

    print("Updated Successfully")
```

```python
def delete_contact():

    cid = int(input("Enter Contact ID: "))

    if cid in contacts:
        del contacts[cid]
        print("Deleted Successfully")
    else:
        print("Contact Not Found")
```

```python


def add_tag():

    cid = int(input("Contact ID: "))

    if cid not in contacts:
        print("Contact Not Found")
        return

    tag = input("Enter Tag: ")

    contacts[cid]["tags"].add(tag)

    print("Tag Added")
```

```python
def remove_tag():

    cid = int(input("Contact ID: "))

    if cid not in contacts:
        print("Contact Not Found")
        return

    tag = input("Enter Tag: ")

    contacts[cid]["tags"].discard(tag)

    print("Tag Removed")
```

```python
def find_by_tag():

    tag = input("Enter Tag: ")

    result = {
        cid: detail
        for cid, detail in contacts.items()
        if tag in detail["tags"]
    }

    if result:

        for cid, detail in result.items():
            print(cid, detail)

    else:
        print("No Contact Found")
```

```python
def save_json():

    try:

        data = {}

        for cid, detail in contacts.items():

            data[cid] = {
                "name": detail["name"],
                "phone": detail["phone"],
                "email": detail["email"],
                "tags": list(detail["tags"]),
                "notes": detail["notes"]
            }

        with open("contacts.json", "w") as file:
            json.dump(data, file, indent=4)

        print("Saved Successfully")

    except Exception as e:
        print(e)
```

```python
def load_json():

    global contacts

    try:

        with open("contacts.json", "r") as file:

            data = json.load(file)

            contacts = {
                int(cid): {
                    "name": detail["name"],
                    "phone": detail["phone"],
                    "email": detail["email"],
                    "tags": set(detail["tags"]),
                    "notes": detail["notes"]
                }
                for cid, detail in data.items()
            }

        print("Loaded Successfully")

    except FileNotFoundError:
        print("JSON File Not Found")

    except json.JSONDecodeError:
        print("Invalid JSON")
```

```python
def statistics():

    print("\nTotal Contacts:", len(contacts))

    all_tags = []

    for detail in contacts.values():
        all_tags.extend(detail["tags"])

    counter = Counter(all_tags)

    print("\nMost Used Tags")

    for tag, count in counter.most_common():
        print(tag, ":", count)

    print("\nContacts Per Tag")

    for tag in counter:
        total = len([
            1
            for detail in contacts.values()
            if tag in detail["tags"]
        ])

        print(tag, ":", total)
```

```python
def sort_contacts():

    sorted_contacts = sorted(
        contacts.items(),
        key=lambda item: item[1]["name"].lower()
    )

    for cid, detail in sorted_contacts:
        print(cid, detail["name"])
```

```python
while True:

    print("\n========== CONTACT MANAGER ==========")

    print("1. Add Contact")
    print("2. Display Contacts")
    print("3. Search Contact")
    print("4. Update Contact")
    print("5. Delete Contact")
    print("6. Add Tag")
    print("7. Remove Tag")
    print("8. Find By Tag")
    print("9. Save JSON")
    print("10. Load JSON")
    print("11. Statistics")
    print("12. Sort Contacts")
    print("13. Exit")

    choice = input("\nEnter Choice: ")

    if choice == "1":
        add_contact()

    elif choice == "2":
        display_contacts()

    elif choice == "3":
        search_contacts()

    elif choice == "4":
        update_contact()

    elif choice == "5":
        delete_contact()

    elif choice == "6":
        add_tag()

    elif choice == "7":
        remove_tag()

    elif choice == "8":
        find_by_tag()

    elif choice == "9":
        save_json()

    elif choice == "10":
        load_json()

    elif choice == "11":
        statistics()

    elif choice == "12":
        sort_contacts()

    elif choice == "13":
        print("Thank You!")
        break

    else:
        print("Invalid Choice")
```
