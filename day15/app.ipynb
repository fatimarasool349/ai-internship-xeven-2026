# 1. LLM   
A Large Language Model (LLM) is an AI model trained on a huge amount of text (books, articles, websites, code, conversations, etc.) to understand and generate human language
## 2. Transformer Architecture Basics
Instead of reading one word after another, they process all words simultaneously
#### Benefits:
- Faster training
- Understands long sentences
- Better context
- More accurate predictions
#### Main Components of a Transformer
**Embedding:** convert text into vector
**Positional Encoding:** Positional encoding tells the model where each word appears.
**Feed Forward Network:** After attention, every word passes through a neural network.
## 3. Attention Mechanism
The model decides which words are most important for understanding another word.
## pre-training vs fine-tuning
|Pre-training|Fine-tuning|
|----|----|
|General knowledge|	Specific task|
|Huge dataset|	Small dataset|
|Learns language|	Learns specialization|
|Very expensive|	Much cheaper|

## Context Window
Context window is the maximum number of tokens the model can consider in one conversation or request.
## Temperature
- Low temperature (e.g., 0–0.3): more deterministic, focused, repeatable outputs — good for factual Q&A, code, classification.
- High temperature (e.g., 0.8–1.2+): more varied, creative, sometimes less coherent outputs — good for brainstorming, creative writing.

## why we need llm ?
1. The Problem with RNN
An RNN (Recurrent Neural Network) reads a sentence one word at a time.
2. Problem: Forgetting Long-Term Information
3. LSTM Improved RNN

An LSTM (Long Short-Term Memory) was designed to reduce this forgetting.
It has special components called gates that decide:

What to remember
What to forget
What to output
4. But LSTM Still Has Problems

Even though LSTMs remember longer context, they still process text one word at a time.

**frequency_penalty:** Reduces repeated words.
**presence_penalty:** Encourages introducing new topics or vocabulary instead of sticking to the same concepts.

## Limitations of LLMs
1. **Hallucinations:** 

Sometimes the model confidently generates information that is incorrect or unsupported.
2. **Knowledge Cutoff:** 

A model's built-in training knowledge only extends up to a certain point in time. Without external tools, it won't know events that happened after that training period.    
3. **Context Limits**:

Even models with large context windows have a limit.
Very long conversations or documents may exceed what fits in the available context.  
4. **Reasoning Errors**

LLMs can make mistakes in:

Complex mathematics
Multi-step logic
Precise counting
Long chains of reasoning


Transformers are better than RNNs and LSTMs because they process all words in parallel instead of one by one. They use self-attention, which lets every word directly relate to every other relevant word, making it easier to understand long-distance relationships in text. This results in faster training, better handling of long contexts, and higher accuracy, which is why modern LLMs use the Transformer architecture.

```python
from openai import OpenAI
from dotenv import load_dotenv
import os

load_dotenv()

client = OpenAI(
    api_key=os.getenv("OPENAI_API_KEY"),
    base_url="https://api.groq.com/openai/v1"

)
```

- client-> object of api 
- .chat-> type of repose of api
- .completion ->  
- create=> sent request to api
- response → the full response returned by Groq.
- choices → a list of generated responses (usually one).
- [0] → take the first response in the list.
- .message → the assistant's message object.
- .content → the actual text of the reply

```python
response = client.responses.create(
    model="llama-3.3-70b-versatile",
    input="What is Artificial Intelligence?",
)

print(response.output_text)


# Print token usage (if available)
print(response.usage)
```

## top_p
top_p controls how many possible next words the AI is allowed to choose from.  
top_p = 0.1 → Only the top 10 words can be used.  
top_p = 0.9 → Hundreds of words are available.  
top_p = Word Choices

## Task 2: Parameter Exploration Notebook 


```python
messages = [
    {
        "role": "system",
        "content": "You are a friendly Python teacher. Explain everything in simple language."
    }
]
while True:
    user_input = input("enter anything that you want to know ")
    if user_input.lower() == "exit":
        break
    messages.append(
    {
        "role": "user",
        "content": user_input
    }
)
```

```python
response = client.chat.completions.create(
    model="llama-3.3-70b-versatile",
    messages=messages,
    temperature=0.7
)
reply = (response.choices[0].message.content)
print("Bot: ", reply)
```
