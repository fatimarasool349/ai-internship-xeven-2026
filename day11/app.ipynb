```python
employee_record =[]
for i in range(1,1001):
    employee = {
        "id": i,
        "name":f"Employee {i}",
        "salary": 30000+ i*300,
    }
    employee_record.append(employee)
print(employee_record)
```

```python
for index, employee in enumerate(employee_record):
    print(index,employee)
```

```python
# apply transformation
for i in range(1,1001,100):
    employee_record[i]["salary"]= None
for i in range(60,1001,100):
    employee_record[i]["salary"]= -90878
for i in range(1,1001,150):
   name = employee_record[i]["name"]
   employee_record[i]["name"]=name.lower()
print(employee_record)
    
```

## Task 2: Pattern Generators 

```python
def table(number):
   for i in range(1,11):
      num= f"{number} * {i}= {number*i}"
      print(num)
number = int(input("Enter a no:"))
table(number)

    
```

```python
# left triangle
for j in range(5):
    stars = (j+1)* "*"
    print(stars)
```

```python
#right triangle
n=5
for j in range(5):
    stars = (n-j)*"*"
    print(stars)
    
  
```

```python
# half paramind
n=5

for j in range(1,6):
    space = (n-j)*" "
    star  = (2*j-1)*"*"
    print(f"{space} {star}")
    
    
```

```python
# 
n=5
for j in range(-5,0):
    space=(n+j)* " "
    star = (-2*j-1)*"*"
    print(f"{space} {star}")
```

```python
# sum rows/columns
matrix = [[3, 6, 9],[1,2,3],[2,4,6]]

print(matrix)
for i in range(3):
    print()
    for j in range(3):
        print(matrix[i][j], end =" ")
        
    
```

```python

for i in range(3):
    sum = 0
    for j in range(3):
        sum += matrix[i][j]
    print(f"{matrix[i]} = {sum}")
```

```python
for i in range(3):
    sum = 0
    for j in range(3):
        sum += matrix[j][i]
    print(f"col {i} sum id {sum}")
```

```python
#transpose
matrix = [[3, 6, 9],[1,2,3],[2,4,6]]

print("Original Matrix:")
for i in range(3):
    print()
    for j in range(3):
        print(matrix[i][j], end =" ")
print("\ntranspose matrix: ")
for i in range(3):
    print()
    for j in range(3):
        print(matrix[j][i], end =" ")
```
