```python
contact = {}
```

```python
def add_contact(id,name,phone,email,tag1,tag2,Note1,Note2):
    if id in contact:
        return "Id already exist"
    contact[id] = {
        "name": name,
        "phone": phone,
        "email": email,
        "tags": {tag1, tag2},      # Set
        "notes": [Note1,Note2]            # List
    }
    return "comtact create sucessful"


add_contact(1,"fatima",923491683632,"fatimarasool@123@gmail.com","friend","forever","i am learning python ","i am try to learn python")
add_contact(2,"ayesha",923491683632,"fatimarasool@123@gmail.com","friend","forever","i am learning python ","i am try to learn python")
print(contact)
        
```

```python
def search_by_name(name):
    
    for id, details in contact.items():
        if details['name'].lower()== name.lower():
            print("name find")
            return details 
    
    
print(contact.items())
name = input("Enter a name you want to search for")
search_by_name(name)
        
    
    
```

```python
def search_by_tag(tag):
    new_list = []
    for id, details in contact.items():
        if tag in details['tags']:
            print("tag find")
            new_list.append(details)
    return new_list
    
    
tag = input("Enter a name you want to search for")
search_by_tag(tag)
        
    
    
```

```python
def add_tag(id, tag):
    for id in  contact:
        id['tags'].add(tag)
    
    
    
id = int(input("Enter the id where you want to add new tag"))
tag = input("enter a tag you want to add")
add_tag(id, tag)
print(contact)
            
        
```
