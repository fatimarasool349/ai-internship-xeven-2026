### 2.1 Data Types — The Four Basic Boxes

Python has four main data types I need to know cold:

- **int** → whole numbers like `5`, `-100`, `0`
- **float** → decimal numbers like `3.14`, `99.99`
- **bool** → only `True` or `False` (capital T and F!)
- **str** → text in quotes like `"hello"` or `'Fatima'`

## Task3: Research 'Python memory management and data types' from 3 AI sources
### Ask ChatGPT: 'How does Python handle memory for different data types?' 
Python stores data as objects in memory and variables act as references to those objects. It automatically allocates memory when objects are created and frees unused memory through reference counting and garbage collection. Different data types (int, float, bool, str) require different amounts of memory based on the data they store.
### Ask Gemini: 'Explain mutable vs immutable types in Python with examples' 
**1. Immutable Types (int, str, tuple)**
Once created, their data cannot change. If you "modify" them, Python builds a new object elsewhere and moves the variable label to it.
**2. Mutable Types (list, dict, set)**
Their internal data can be changed directly inside the original memory container.

### Ask Claude: 'What are best practices for naming variables in Python?' 
Follow PEP 8 — the official Python style guide:

- Use snake_case:  all lowercase, words separated by underscores
- Be descriptive: user_age not x, monthly_salary not ms
- Constants in ALL_CAPS: MAX_RETRIES = 3, PI = 3.14
- Classes use PascalCase: class StudentRecord

Length rule of thumb:

- Short scope (loop counter): short name is fine: for i in range(10)
- Wide scope (used across file): be explicit: total_invoice_amount
