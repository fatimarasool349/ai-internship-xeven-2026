## Tuples
A tuple is an ordered, immutable sequence of elements. Once created, its contents cannot be changed (no adding, removing, or modifying elements).
### Why are tuples faster than lists?
Python does not need to allocate extra memory for resizing or modifying them because tuple does not changeable.
### Sets 
A set is an unordered collection of unique elements. no indexing
#### Why Use Sets?

Sets are useful when:

- Removing duplicates
- Fast searching
- Mathematical set operations

## task2: Unique Visitor Tracker 


```python
day1= ["192.0.0.3","194.0.0.1","10.0.0.1","10.0.0.1"]
day2 = ["192.0.0.3","192.0.0.3","10.0.0.2","10.0.0.9"]


unique_ip_address_day_one = set(day1)
print(unique_ip_address_day_one)

unique_ip_address_day_two = set(day2)
print(unique_ip_address_day_two)

union = unique_ip_address_day_one.union(unique_ip_address_day_two)
print(union)
```

```python

difference_day_one = unique_ip_address_day_one.difference(unique_ip_address_day_two)
print(difference_day_one)

difference_day_two = unique_ip_address_day_two.difference(unique_ip_address_day_one)
print(difference_day_two)

union_of_both_difference = difference_day_one.union(difference_day_two)
print(union_of_both_difference)

common_visitor = unique_ip_address_day_one.intersection(unique_ip_address_day_two)
print(common_visitor)
```

```python
growth_rate = ((len(unique_ip_address_day_two) - len(unique_ip_address_day_one))
               / len(unique_ip_address_day_one)) * 100
print(f"growth_rate of date one", growth_rate )
retention_rate = (len(common_visitor)/len(unique_ip_address_day_one))*100
print(f"retention rate { retention_rate}")
```
